```python
import numpy as np
import matplotlib.pyplot as plt

def sigmoid(x):
    return 1/(1+np.exp(-x))

def tanh(x):
    return (np.exp(x) - np.exp(-x)) / (np.exp(x) + np.exp(-x))

def relu(x):
  return np.maximum(0,x)

def leaky_relu(x):
  return np.maximum(0.01*x,x)

def softmax(x):
  return np.exp(x)/np.sum(np.exp(x))

x=np.linspace(-10,10,100)

plt.figure(figsize=(10,5))

plt.subplot(3,2,1)
plt.plot(x,sigmoid(x))
plt.title("sigmoid")
plt.grid(True)

plt.subplot(3,2,2)
plt.plot(x,tanh(x))
plt.title("tanh")
plt.grid(True)

plt.subplot(3,2,3)
plt.plot(x,relu(x))
plt.title("relu")
plt.grid(True)

plt.subplot(3,2,4)
plt.plot(x,leaky_relu(x))
plt.title("leaky relu")
plt.grid(True)

plt.subplot(3,2,5)
plt.plot(x,softmax(x))
plt.title("softmax")
plt.grid(True)

plt.tight_layout()
plt.show()
```
